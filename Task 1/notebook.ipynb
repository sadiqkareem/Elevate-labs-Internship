///Converting categorical data into numerical using encoder///


```python
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
df.Name =  le1.fit_transform(df.Name)
le2 = LabelEncoder()
df.Sex =  le2.fit_transform(df.Sex)
le3 = LabelEncoder()
df.Embarked =  le3.fit_transform(df.Embarked)
le4 = LabelEncoder()
df.Ticket =  le4.fit_transform(df.Ticket)
le5 = LabelEncoder()
df.Cabin =  le5.fit_transform(df.Cabin)
```

///Nomralize the numerical features///

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Select numerical columns
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns

# Standardize the numerical features
df[numerical_features] = scaler.fit_transform(df[numerical_features])

print(df.head())
```

Removing outliers and visualizing the data using boxplots

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select numeric columns
numeric_cols = df.select_dtypes(include='number').columns

# 1. Visualize outliers before removal
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[numeric_cols])
plt.xticks(rotation=45)
plt.title("Boxplots Before Outlier Removal")
plt.show()

# 2. Calculate IQR and remove outliers
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

# Keep rows where every numeric column is within 1.5 × IQR
df_clean = df[
    ~((df[numeric_cols] < (Q1 - 1.5 * IQR)) |
      (df[numeric_cols] > (Q3 + 1.5 * IQR))).any(axis=1)
]

# 3. Visualize after removal
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean[numeric_cols])
plt.xticks(rotation=45)
plt.title("Boxplots After Outlier Removal")
plt.show()

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Rows removed:", len(df) - len(df_clean))
```
